```python
import pandas as pd
import matplotlib.pyplot as plt

trainPath1 = './800_Testbed/800_0/800_0_1.dat'
trainPath2 = './800_Testbed/800_0/800_0_2.dat'

loaded_train1 = pd.read_csv(trainPath1,delimiter='\t')['v1'].values
loaded_train2 = pd.read_csv(trainPath2,delimiter='\t')['v1'].values

faultPath = './800_Testbed/800_0_BFI/800_0_BFI.dat'
faultLoad = pd.read_csv(faultPath,delimiter='\t')['v1'].values

plt.plot(range(len(loaded_train1)),loaded_train1,'b')
plt.plot(range(len(faultLoad)),faultLoad,'r')
plt.show()
```

```python
train_path_01 = './800_Testbed/pkled_data/noised_data/800_0_interval_4096_noiseRatio_0.2/trainVal_800_0_interval_4096_noised_0.2.pkl'
test_path_01 = './800_Testbed/pkled_data/noised_data/800_0_interval_4096_noiseRatio_0.2/test_800_0_interval_4096_noised_0.2.pkl'

import pickle
def open_pickle(pkl_path):
    with open(pkl_path,'rb') as F:
        loaded_pkl = pickle.load(F)
        
    return loaded_pkl

train_data_01 = open_pickle(train_path_01)
test_data_01 = open_pickle(test_path_01)

import matplotlib.pyplot as plt

for i in train_data_01:
    
    if i[1] == 0: 
        plt.plot(
            range(len(i[0])),
            i[0],
            'b',
            alpha=0.01
        )
    elif i[1] == 1:
        plt.plot(
            range(len(i[0])),
            i[0],
            'r',
            alpha=0.1
        )
    else:
        raise Exception('wrong')
plt.show()
plt.close()

for i in test_data_01:
    
    if i[1] == 0: 
        plt.plot(
            range(len(i[0])),
            i[0],
            'b',
            alpha=0.01
        )
    elif i[1] == 1:
        plt.plot(
            range(len(i[0])),
            i[0],
            'r',
            alpha=0.1
        )
    else:
        raise Exception('wrong')
plt.show()
    
```
