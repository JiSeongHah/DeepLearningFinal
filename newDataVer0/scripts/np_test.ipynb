```python
num_lst = [
    947,
    474,
    237,
    118
]

ratio_lst = [1,2,4,8,16]
import random

num_to_name = {
    947 : '512',
    474 : '1024',
    237 : '2048',
    118 : '4096',
}
total_dict = {
    '512': {},
    '1024':{},
    '2048':{},
    '4096':{},
}

for num in num_lst:
    
    for ratio in ratio_lst:
        
        slice_num = round(num/ratio)
        
        random_pick = random.sample(
            [i for i in range(num)],
            slice_num
        )
        
        bool_lst = []
        for i in range(num):
            if i in random_pick:
                bool_lst.append(True)
            else:
                bool_lst.append(False)
        
        total_dict[
            num_to_name[num]
        ][ratio] = bool_lst
        
        print(total_dict[
            num_to_name[num]
        ][ratio])
        print(len(total_dict[
            num_to_name[num]
        ][ratio]))
    
    print(total_dict)
        
        
    print('----------')

import pickle  
with open('./mask_per_ratio.pkl','wb') as F:
    pickle.dump(total_dict,F)
    
print('mission_complete')
```
