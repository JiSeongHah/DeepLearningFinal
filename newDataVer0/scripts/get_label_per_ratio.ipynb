```python
import pickle 
import numpy as np

with open('./label_only_results_image_ver.pkl','rb') as F:
    label_only_image_ver_dict = pickle.load(F)


label_ratio_lst = [round(0.1*i,1) for i in range(1,11)]
label_per_ratio_dict = {}
for k,v in label_only_image_ver_dict.items():
    
    normal_num = len(v[v == 0])
    print(normal_num,len(v[v==1]))
    label_per_ratio_dict[k] = {}
    for label_ratio in label_ratio_lst:
        
        choice_num = round(
            normal_num*label_ratio
        )
        
        print(label_ratio,choice_num)
        
        normal_idx_arr = np.where(v == 0)[0]
        
        sampled_normal_idx_arr = np.random.choice(
                normal_idx_arr,
                size=choice_num,
                replace=False
            )
        abnormal_idx_arr = np.where(
            v == 1
        )[0]
        
        filtered_idx_arr = np.sort(
            np.concatenate(
                [
                    sampled_normal_idx_arr,
                    abnormal_idx_arr
                ]
            )
        )
        
        new_key = 'normal_label_filter_ratio'+'_'+str(label_ratio)
        
        print(new_key,len(filtered_idx_arr))
        print(filtered_idx_arr)
        label_per_ratio_dict[k][new_key] = filtered_idx_arr
        
        print(v[label_per_ratio_dict[k][new_key]].shape,len(v[label_per_ratio_dict[k][new_key]][v[label_per_ratio_dict[k][new_key]]==1]),len(v[label_per_ratio_dict[k][new_key]][v[label_per_ratio_dict[k][new_key]]==0]))
        print('\n')
with open('./label_idx_arr_per_ratio_image_ver.pkl','wb') as F:
    pickle.dump(label_per_ratio_dict,F)
    
print('mission complete')
        
        
        
        
        
#     print()
        
        
        
# x = np.array([
#     i*i for i in range(1,21)
# ])   
# print(x)
# y = np.random.randint(0,10,[20])
# print(y)
# z = np.where(y>5)
# print(z)
# zz = np.sort(np.random.choice(z[0],size=5,replace=False))
# print(zz,type(zz))
# print(x[zz])

# zzz =np.where(y<=5)[0]
# print(zzz.shape)
# zzzz = np.concatenate(
#     [
#         zz,
#         zzz
#     ]
# )

# print(zzzz)
    


    
    
    
```
