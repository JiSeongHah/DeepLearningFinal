```python
import numpy as np




class class_FeatureExtractionFreq():
    
    def __init__(self):
    
        self.EPS = 1e-8
        
#         print('updated freq')
    
    def getMean(self,AmpArr):

        return np.mean(AmpArr)

    def getVarOfMean(self,AmpArr):

        return np.var(AmpArr,ddof =1)


    def getSkewnessPowerSpectrum(self,AmpArr):

        mean = np.mean(AmpArr)

        var = self.getVarOfMean(AmpArr)


        return np.mean((AmpArr-mean)**3)/((np.sqrt(var)**3))

    def getKurtosisPowerSpectrum(self,AmpArr):

        mean = np.mean(AmpArr)

        var = self.getVarOfMean(AmpArr)

        return np.mean((AmpArr-mean)**4)/((var**2))

    def getFrequencyCentre(self,FreqArr,AmpArr):

        return np.sum(FreqArr*AmpArr)/(np.sum(AmpArr))

    def getRootVar(self,FreqArr,AmpArr):

        freqCentre = self.getFrequencyCentre(FreqArr,AmpArr)

        return np.sqrt(np.mean(((FreqArr-freqCentre)**2)*AmpArr))

    def getRMS(self,FreqArr,AmpArr):

        numer = np.sum((FreqArr**2)*AmpArr)

        denom = np.sum(AmpArr)

        return np.sqrt(numer/(denom))

    def getMeanFreqOverMeanTime(self,FreqArr,AmpArr):

        numer = np.sum((FreqArr**4)*AmpArr)

        denom = np.sum((FreqArr**2)*AmpArr)

        return np.sqrt(numer/(denom))


    def getStabilisationFactor(self,FreqArr,AmpArr):

        numer = np.sum((FreqArr**2)*AmpArr)

        denom = np.sqrt(np.sum(AmpArr)*np.sum((FreqArr**4)*AmpArr))

        return numer / (denom)


    def getCoeffOfVariability(self,FreqArr,AmpArr):

        return self.getRootVar(FreqArr,AmpArr)/(self.getFrequencyCentre(FreqArr,AmpArr))


    def getSkewness(self,FreqArr,AmpArr):

        numer = np.mean(((FreqArr-self.getFrequencyCentre(FreqArr,AmpArr))**3)*AmpArr)

        denom = self.getRootVar(FreqArr,AmpArr)**3

        return numer/(denom)

    def getKurtosis(self,FreqArr,AmpArr):

        numer = np.mean(((FreqArr-self.getFrequencyCentre(FreqArr,AmpArr))**4)*AmpArr)

        denom = self.getRootVar(FreqArr,AmpArr)**4

        return numer/(denom)


    def getRMSRatio(self,FreqArr,AmpArr):

        numer = np.mean(np.sqrt(np.abs((FreqArr-self.getFrequencyCentre(FreqArr,AmpArr))))*AmpArr)
#         numer = np.mean(np.sqrt(FreqArr-self.getFrequencyCentre(FreqArr,AmpArr))*AmpArr)

        denom = np.sqrt(self.getRootVar(FreqArr,AmpArr))

        return numer/(denom)
    
    
    def getFeatureExtractionFreq(self,FreqArr,AmpArr):



        return np.array([self.getMean(AmpArr),
                         self.getVarOfMean(AmpArr),
                         self.getSkewnessPowerSpectrum(AmpArr),
                         self.getKurtosisPowerSpectrum(AmpArr),
                         self.getFrequencyCentre(FreqArr,AmpArr),
                         self.getRootVar(FreqArr,AmpArr),
                         self.getRMS(FreqArr,AmpArr),
                         self.getMeanFreqOverMeanTime(FreqArr,AmpArr),
                         self.getStabilisationFactor(FreqArr,AmpArr),
                         self.getCoeffOfVariability(FreqArr,AmpArr),
                         self.getSkewness(FreqArr,AmpArr),
                         self.getKurtosis(FreqArr,AmpArr),
                         self.getRMSRatio(FreqArr,AmpArr)
                         ])

```
