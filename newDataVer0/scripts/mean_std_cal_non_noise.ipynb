```python
import torch
import numpy as np
from torchvision import datasets,transforms
import os
import pickle 

save_path = './data_download_path/non_noise_data/mnist_ver/mnist_processed_train.pkl'
with open(save_path,'rb') as F:
    loaded_data = pickle.load(F)
    
data_per_label_dict= {i:[] for i in range(10)}
for each_data in loaded_data:
    
    each_input,each_label = each_data[0], each_data[1]
    
    data_per_label_dict[each_label].append(each_input)
    
    
mean_std_per_label_dict={str(i):{} for i in range(10)}
for k,v in data_per_label_dict.items():
    
    data_per_label_dict[k] = torch.stack(v)
    
    
for k,v in data_per_label_dict.items():
    
    each_mean = torch.mean(v).item()
    each_std = torch.std(v).item()
    
    mean_std_per_label_dict[str(k)]['mean'] = each_mean
    mean_std_per_label_dict[str(k)]['std'] = each_std
    
with open('./data_download_path/non_noise_data/mnist_ver/configs/mean_std_mnist_train.pkl','wb') as F:
    pickle.dump(mean_std_per_label_dict,F)
    
print('done')
    
    
    
    
    
    
    
```
