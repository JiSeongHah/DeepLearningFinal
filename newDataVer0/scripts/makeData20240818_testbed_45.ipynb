```python
def openPickle(loadPath):
    
    with open(loadPath, 'rb') as f:
        data = pickle.load(f)
        
        return data
    
def savePickle(savePath,data):
    with open(savePath, 'wb') as f:
        pickle.dump(data, f, pickle.HIGHEST_PROTOCOL)
        
    print(f'saving data into {savePath} complete')

import numpy as np
from copy import deepcopy

def saveMeanStd(dataType,interval,noiseRatio):
    
    assert dataType in ['800_0','800_45']
    
    trainPath = f'./800_Testbed/pkled_data/noised_data/{dataType}_interval_{interval}_noiseRatio_{noiseRatio}/trainVal_{dataType}_interval_{interval}_noised_{noiseRatio}.pkl'
    
    trainData = openPickle(trainPath)
    
    newTrainData = []
    for eachTrainData in trainData:
        if eachTrainData[2] == 'train':
            newTrainData.append(
                eachTrainData[0]
            )
        
        
    newTrainData = np.concatenate(newTrainData)
    
    meanValue = np.mean(newTrainData)
    stdValue = np.std(newTrainData,ddof=1)
    
    saveDict = {
        'mean': meanValue,
        'std': stdValue
    }
    
    savePickle(
        f'./800_Testbed/pkled_data/noised_data/{dataType}_interval_{interval}_noiseRatio_{noiseRatio}/trainVal_{dataType}_interval_{interval}_noised_{noiseRatio}_mean_std.pkl',
        saveDict
    )
    print(f'./800_Testbed/pkled_data/noised_data/{dataType}_interval_{interval}_noiseRatio_{noiseRatio}/trainVal_{dataType}_interval_{interval}_noised_{noiseRatio}_mean_std.pkl', 'done')
    
    
    
    
        
dataType_lst = ['800_0','800_45']
intervalLst = [128,256,512,1024,2048,4096]
noiseRatio_lst = [round(i*0.1,1) for i in range(0,10)]


for dataType in dataType_lst:
    for interval in intervalLst:
        for noiseRatio in noiseRatio_lst:
            saveMeanStd(
                dataType=dataType,
                interval=interval,
                noiseRatio=noiseRatio
            )
            
        
        
```
