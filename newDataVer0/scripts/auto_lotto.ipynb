```python
import random
import numpy as np

exclude_lst = []
choice_lst = []

def return_num(exclude_nums,choice_lst,first_num):
    
    # num_lst_10 = [i for i in range(1,1+10)]
    # num_lst_20 = [i for i in range(11,11+10)]
    # num_lst_30 = [i for i in range(21,21+10)]
    # num_lst_40 = [i for i in range(31,31+10)]
    # num_lst_50 = [i for i in range(41,41+10)]
    
    num_lst = [i for i in range(1,1+45) if i not in exclude_nums]
    num_lst = [i for i in num_lst if i>=first_num]
    

    num_lst_10 = []
    num_lst_20 = []
    num_lst_30 = []
    num_lst_40 = []
    num_lst_50 = []
    
    for num in num_lst:
        
        if num >=1 and num < 11:
            num_lst_10.append(num)
        elif num >=10 and num < 20:
            num_lst_20.append(num)
        elif num >=20 and num < 30:
            num_lst_30.append(num)
        elif num >=30 and num < 40:
            num_lst_40.append(num)
        elif num >=40 and num < 46:
            num_lst_50.append(num)
            
    total_lst = [
        num_lst_10,
        num_lst_20,
        num_lst_30,
        num_lst_40,
        num_lst_50,
    ]
    
    final_choice = []
    
    for refined_lst,how_many_num in zip(total_lst,choice_lst):
        # print(refined_lst,how_many_num)
        if first_num in refined_lst:
            samples = [first_num]+random.sample(refined_lst,how_many_num-1)
        else:
            samples = random.sample(refined_lst,how_many_num)
        
        final_choice.extend(samples)
    
    return sorted(final_choice)


        

for i in range(5):
    result = return_num(
        exclude_nums=[8,15,19,21,32,36,38],
        choice_lst=[2,1,2,1,0],
        first_num=4
    )
    print(result)
        
for i in range(5):
    result = return_num(
        exclude_nums=[8,15,19,21,32,36,38],
        choice_lst=[1,2,2,1,0],
        first_num=5
    )
    print(result)
        



```
